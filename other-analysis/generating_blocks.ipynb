```python
def cross_group_combinations(lang_dict, exclude_key):
    """Returns all unique combinations of languages across different keys in lang_dict, excluding the specified key."""
    result = []

    # Collect keys except the excluded one
    remaining_keys = [k for k in lang_dict if k != exclude_key]

    # For each pair of different keys, cross their languages
    for i in range(len(remaining_keys)):
        for j in range(i + 1, len(remaining_keys)):
            key_a = remaining_keys[i]
            key_b = remaining_keys[j]
            for lang_a in lang_dict[key_a]:
                for lang_b in lang_dict[key_b]:
                    result.append((lang_a, lang_b))

    return result
```

#### Generate blocks for 1A
- Only for is_shown == 1
- First langauge is the language itself 
- Second language is the other language in its group where is_shown == 1
- Third language is any language where is_shown == 1 and group != same as first language
- Fourth language is any language where is_shown == 1 and group != same as first or group != same as thirds

```python
# Below are the possible groups:
# Group1: Achuar, Aguaruna
# Group2: Lithuanian, Croatian
# Group3: Dutch, Norwegian
# Group4: Wolof, Indonesian

# For each language give me the following: a list of their language group and then a random language from each other group, we need each permutation


langs = {1: ["Achuar", "Aguaruna"],
         2: ["Lithuanian", "Croatian"],
         3: ["Dutch", "Norwegian"],
         4: ["Wolof", "Indonesian"]}

combinations = {}

for key in langs:
    outside_combinations = cross_group_combinations(langs, key)
    for lang in langs[key]:
        first = lang
        second = [l for l in langs[key] if l != lang][0]  # The other language in the same group
        combinations[lang] = random.sample([[first, second, third, fourth]  for third, fourth in outside_combinations], 12)

# save in language_options
for lang in combinations:
    if lang in combinations:
        language_options[lang]["1A"] = combinations[lang]

language_options
```

### Generate blocks for question 2

```python
# example [1.1, 1.2, 2.1, 3.2] where X.Y means language Y from group X
# in the 1.X pair we have the following combinations for 3rd and 4th lang: 2.1:3.1, 2.1:3.2, 2.2:3.1, 2.2:3.2, 2.1:4.1, 2.1:4.2, 2.2:4.1, 2.2:4.2, 3.1:4.1, 3.1:4.2, 3.2:4.1, 3.1:4.2
# so 12 combinations per pair, we have 4 different pairs in each group so total permutations is 4 * 12
n_combinations = 48

langs = {1: ["Achuar", "Aguaruna"],
         2: ["Lithuanian", "Croatian"],
         3: ["Dutch", "Norwegian"],
         4: ["Wolof", "Indonesian"]}

combinations = {1: [], 2: [], 3: [], 4: []}

cnt = 0
for key in langs:
    outside_combinations = cross_group_combinations(langs, key)
    first = langs[key][0]
    second = langs[key][1]
    combinations_entry = [random.sample([first, second] + [third, fourth], 4)  for third, fourth in outside_combinations]
    combinations[key].extend(combinations_entry)
    cnt += len(combinations_entry)
        
print(f'Total combinations generated: {cnt}, expected: {n_combinations}')  # Should be 48
combinations
```

```python
def get_n_combinations(combinations, n=20):
    """Returns a list of n sets of combinations, each set containing 10 combinations (as actual combination lists)."""
    flat_combinations = [item for sublist in combinations.values() for item in sublist]
    global_idx_map = {tuple(sorted(item)): i for i, item in enumerate(flat_combinations)}
    indice_counts = {i: 0 for i in range(len(flat_combinations))}
    all_combinations = []
    for _ in range(n):
        selected_items = []  # store actual combinations (lists of languages)
        picked_set = set()  # indices already picked for this 10-set

        # 1) For each group: pick the two least-picked combinations (ties broken randomly)
        for key in combinations:
            group_items = combinations[key]
            # translate group items to indices (use normalized tuple key)
            group_indices = []
            for item in group_items:
                k = tuple(sorted(item))
                group_indices.append(global_idx_map[k])

            # sort by current count, break ties randomly
            ranked = sorted(group_indices, key=lambda idx: (indice_counts.get(idx, 0), random.random()))
            for idx in ranked[:2]:
                picked_set.add(idx)
                indice_counts[idx] = indice_counts.get(idx, 0) + 1
                selected_items.append(flat_combinations[idx])

        # 2) From the remaining global pool, pick the two least-picked (ties broken randomly)
        remaining_indices = [i for i in range(len(flat_combinations)) if i not in picked_set]
        remaining_sorted = sorted(remaining_indices, key=lambda idx: (indice_counts.get(idx, 0), random.random()))
        for idx in remaining_sorted[:2]:
            picked_set.add(idx)
            indice_counts[idx] = indice_counts.get(idx, 0) + 1
            selected_items.append(flat_combinations[idx])

        all_combinations.append(selected_items)

    return all_combinations
```

```python
for lang in language_options:
    combs = get_n_combinations(combinations, n=int(quotas[lang]["2A"]["quota"]))
    language_options[lang]["2A"] = combs
```

#### Small sanity check for balance

```python
# look at distribution of combinations in trial (now trial contains actual combination lists, not indices)
trial = get_n_combinations(combinations, n=70)


# Count how often each global combination appears by comparing lists
combination_counts = {i: 0 for i in range(48)}
combination_lookup = {}
for trial_set in trial:
    for comb in trial_set:
        if tuple(sorted(comb)) in combination_lookup:
            idx = combination_lookup[tuple(sorted(comb))]
            combination_counts[idx] += 1
        else:
            # register new idx for this combination
            idx = len(combination_lookup)
            combination_lookup[tuple(sorted(comb))] = idx
            combination_counts[idx] = 1

sns.scatterplot(list(combination_counts.values()))
```

```python
# combine combination_lookup and combination_count, where key = combination tuple and value = count
combination_distribution = {comb: combination_counts[idx] for comb, idx in combination_lookup.items()}

# now visualise distribution of combinations, y-axis = combination, x-axis = count
comb_df = pd.DataFrame(list(combination_distribution.items()), columns=["Combination", "Count"])

import matplotlib.pyplot as plt
plt.figure(figsize=(10, 16))
ax = sns.barplot(x=comb_df["Count"].values, y=[str(comb) for comb in comb_df["Combination"].values], orient='h', color='#1f77b4', order=comb_df.sort_values("Count", ascending=False)["Combination"].apply(str))
ax.set_title(f'Distribution of combinations across {len(trial)} trial sets')
ax.set_xlabel('Count')
ax.set_ylabel('Combination')
```
